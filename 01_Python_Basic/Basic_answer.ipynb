**Write a Program to find the largest of three numbers.**

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))

if a > b and a > c:
    print(f"{a} is greater.")
elif b > a and b > c:
    print(f"{b} is greater.")
else:
    print(f"{c} is greater.")
```

**Write a Python program to calculate the factorial of a number.**

```python
fact = 1
num = int(input("Enter a number: "))

for i in range(1,num+1):
    fact = fact * i
print(fact)
```

**Write a Python program to swap the values of two variables.**

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
print("Before Swapping")
print(a)
print(b)
print("After Swapping")
a, b = b, a
print(a)
print(b)
```

**Write a Python program to find the sum of first n natural numbers.**

```python
num = int(input("Enter a number: "))

total = 0

for i in range(num+1):
    total = total + i
print(f"The Sum of {num} numbers: {total}")
```

**Write a Python program to check if a year is a leap year.**

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print("Leap Year")
else:
    print("Not a leap year.")
```

**Write a Python program to check if a string is a palindrome.**

```python
name = input("Enter any  character: ")

if name == name[::-1]:
    print(f"It is a Palindrome.")
else:
    print(f"It is not a Palindrome.")
```
